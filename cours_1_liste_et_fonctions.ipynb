# Exercice. 3. Écris compter_mot(chemin, mot) qui renvoie le nombre de lignes contenant mot

```python
# les lignes qui contiennent le mot
def compter_mot(chemin, mot):
    n = 0
    with open(chemin, encoding="utf-8") as f:
        for ligne in f:
            if mot.lower() in ligne.lower():
                n += 1
    return n

print(compter_mot("test.txt", "python"))  # Affiche lignes contenant "python" dans le fichier test.txt
```

# Séance 2 : dictionnaires, sets et slicing
Les notions

```python
# dict : paires clé -> valeur
ages = {"Awa": 31, "Koffi": 28}
ages["Yao"] = 40                  # ajout
ages.get("Zoe", 0)                # 0 (pas d'erreur si absente)
for nom, age in ages.items():     # parcours clé + valeur
    print(nom, age)

# set : éléments uniques, test d'appartenance rapide
s = set([1, 2, 2, 3, 3])          # {1, 2, 3}
2 in s                            # True

# slicing : [début:fin:pas], fin exclue
l = [10, 20, 30, 40, 50]
l[1:4]      # [20, 30, 40]
l[:2]       # [10, 20]
l[::-1]     # [50, 40, 30, 20, 10]
"python"[:3]   # "pyt"
```

# Ex. 5. Écris compter_lettres(texte) qui renvoie un dict du nombre d’occurrences de chaque lettre.

```python
def compter_lettres(texte):
    # Initialisation du dictionnaire des fréquences
    occurrences = {}
    
    # Parcours de chaque caractère du texte
    for caractere in texte.lower():
        if caractere.isalpha():  # On ne garde que les lettres
            if caractere in occurrences:
                occurrences[caractere] += 1
            else:
                occurrences[caractere] = 1
                
    return occurrences
print(compter_lettres("banane"))
print(compter_lettres("Bonjour tout le monde !"))
```

```python
#Une version plus courte

# le if / else peut se condenser avec .get(), vu dans la séance :

def compter_lettres(texte):
    occurrences = {}  # Initialisation du dictionnaire des fréquences
    for c in texte.lower():
        if c.isalpha():  # On ne garde que les lettres
            occurrences[c] = occurrences.get(c, 0) + 1
    return occurrences

print(compter_lettres("banane"))
print(compter_lettres("Bonjour tout le monde !"))
```

# Exercice 6: doublons(liste), qui renvoie {2, 3} pour [1, 2, 2, 3, 3, 3, 4]. Utilise deux set : 
un pour ce qu’on a déjà vu, un pour les doublons.

```python
def doublons(liste):
    deja_vu = set()
    les_doublons = set()
    
    for element in liste:
        if element in deja_vu:
            les_doublons.add(element)
        else:
            deja_vu.add(element)
            
    return les_doublons

# Exemple d'utilisation :
test_liste = [1, 2, 2, 3, 3, 3, 4]
print(doublons(test_liste))  # Sortie : {2, 3}

# Tests supplémentaires pour vérifier le comportement de la fonction doublons
print(doublons([]))              # set()   (et non {} qui est un dict vide !)
print(doublons([1, 2, 3]))       # set()
print(doublons(["a", "b", "a"])) # {'a'}
```

# Ex. 7 : est_palindrome(mot) en une ligne, avec le slicing [::-1].
Un palindrome est un mot, une phrase ou un nombre qui se lit de la même manière dans les deux sens (de gauche à droite et de droite à gauche).
Voici quelques exemples simples :
• Mots : radar, kayak, rotor, été, ici, elle, laval.
• Nombres : 121, 1331, 2002.


```python
def est_palindrome(mot):
    for c in mot:
        if not c.isalpha():
            return False
    return mot.lower() == mot.lower()[::-1]

# Exemples d'utilisation :
print(est_palindrome("radar"))  # Sortie : True
print(est_palindrome("python")) # Sortie : False
print(est_palindrome("Esope reste ici et se repose"))  # Sortie : True
# Alternative de la fonction est_palindrome pour plus de clarté
#def est_palindrome(mot):
    #mot = mot.lower()
    #return mot == mot[::-1]
```

Voici comment généraliser la fonction pour qu'elle devienne insensible aux espaces, aux majuscules et aux accents en combinant une compréhension de liste et "".join() :

```python
def est_palindrome_general(texte):
    # 1. Ne garder que les lettres et tout mettre en minuscules
    nettoye = "".join([c for c in texte.lower() if c.isalpha()])
    
    # 2. Vérifier si le texte nettoyé est égal à son inverse
    return nettoye == nettoye[::-1]

# Test avec ta phrase :
print(est_palindrome_general("Esope reste ici et se repose"))  # Sortie : True
```

# Les slicing

```python
# Prenons la chaîne : mot = "PYTHON"

#1. Extraire une partie (début et fin)
mot = "PYTHON"
print(mot[0:3])  # Sortie : "PYT" (indices 0, 1, 2. L'indice 3 'H' est exclu)
print(mot[2:5])  # Sortie : "THO" (indices 2, 3, 4)
```

```python

# 2. Raccourcis (omettre des valeurs)

mot = "PYTHON"
print(mot[:4])   # Sortie : "PYTH" (du début jusqu'à l'indice 4 exclu)
print(mot[2:])   # Sortie : "THON" (de l'indice 2 jusqu'à la fin)
print(mot[:])    # Sortie : "PYTHON" (copie entière de la séquence)
```

```python

# 3. Utiliser le pas

print(mot[0:6:2]) # Sortie : "PTO" (indices 0, 2, 4)
```

```python

#Le cas particulier de [::-1] (Inverser)

#Dans l'exercice du palindrome, nous avons utilisé mot[::-1]. Voici comment Python l'interprète :
#• début et fin sont vides : on prend tout le mot.
#• pas vaut -1 : on parcourt la chaîne à l'envers, un caractère à la fois.
mot = "PYTHON"
print(mot[::-1])  # Sortie : "NOHTYP"
```
