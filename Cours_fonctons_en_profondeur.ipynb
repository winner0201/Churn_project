# Séance 3 : les fonctions en profondeur
Les notions

```python
# 1. Arguments nommés et valeurs par défaut
def saluer(nom, politesse="Bonjour"):
    return f"{politesse} {nom}"

saluer("Awa")                        # "Bonjour Awa"
print(saluer("Awa", politesse="Salut") )    # "Salut Awa"

# 2. *args : nombre variable d'arguments (reçus sous forme de tuple)
def somme(*nombres):
    return sum(nombres)

print(somme(1, 2, 3) )                      # 6

# 3. **kwargs : arguments nommés variables (reçus sous forme de dict)
def afficher(**infos):
    for cle, valeur in infos.items():
        print(cle, "=", valeur)

print(afficher(nom="Awa", age=31))

# 4. Portée : une variable créée dans une fonction n'existe que dedans
def f():
    x = 10
f()
print(x)    # NameError
```

# Ex. 8. Écris moyenne(*nombres) qui renvoie la moyenne des nombres reçus, et 0 s’il n’y en a aucun.

```python
def moyenne(*nombres):
    if not nombres:
        return 0
    return sum(nombres) / len(nombres)

print(moyenne(10, 20, 30))   # 20.0
print(moyenne())             # 0
```

# Ex. 9. Écris creer_profil(nom, **infos) qui renvoie un dict contenant le nom et toutes les infos supplémentaires.

```python

def creer_profil(nom, **infos):
    profil = {"nom": nom}
    profil.update(infos)
    return profil

print(creer_profil("Awa", age=31, ville="Paris", code_postal="75000"))  # {"nom": "Awa", "age": 31, "ville": "Paris", "code_postal": "75000"}
# {"nom": "Awa", "age": 31, "ville": "Paris"}
```

# Séance 4 : exceptions et fichiers
Les notions

```python
# 1. try / except : gérer une erreur au lieu de planter
try:
    nombre = int("abc")
except ValueError:
    print("Ce n'est pas un nombre")

# 2. Plusieurs cas, else et finally
try:
    f = open("absent.txt", encoding="utf-8")
except FileNotFoundError:
    print("Fichier introuvable")
else:
    print("Ouverture réussie")    # seulement s'il n'y a PAS eu d'erreur
    f.close()
finally:
    print("Toujours exécuté")     # dans tous les cas

# 3. Lever sa propre exception
def racine(x):
    if x < 0:
        raise ValueError("x doit être positif")
    return x ** 0.5

# 4. pathlib : manipuler les chemins proprement
from pathlib import Path
p = Path("test.txt")
p.exists()            # True / False
p.read_text(encoding="utf-8")
p.write_text("Bonjour", encoding="utf-8")
```

# Ex. 11. Écris convertir_en_entier(texte) qui renvoie l’entier correspondant, ou None si la conversion est impossible.

```python
def convertir_en_entier(texte):
    try:
        return int(texte)
    except (ValueError, TypeError):
        return None

if convertir_en_entier("abc") is None:
    print("Conversion impossible")
    
print(convertir_en_entier("42"))     # 42
print(convertir_en_entier("abc"))    # None
print(convertir_en_entier(""))       # None
```

# Ex. 12. Écris lire_premiere_ligne(chemin) qui renvoie la première ligne du fichier, ou le texte "Fichier introuvable" si le fichier n’existe pas.

```python
def lire_premiere_ligne(nom_fichier):
    try:
        with open(nom_fichier, encoding="utf-8") as f:
            return f.readline().strip()
    except FileNotFoundError:
        return "Fichier introuvable"
print(lire_premiere_ligne("test.txt"))      # "Bonjour le monde"
print(lire_premiere_ligne("absent.txt"))    # "Fichier introuvable"
```

# Ex. 13. Écris diviser(a, b) qui lève une ValueError avec le message "Division par zéro interdite" si b vaut 0, sinon renvoie a / b. Teste-la ensuite avec un try / except qui affiche le message d’erreur.

```python
def diviser(a, b):
    if b == 0:
        raise ValueError("Division par zéro interdite")
    return a / b

# Le try / except se place à l'endroit de l'APPEL
try:
    print(diviser(10, 2))    # 5.0
    print(diviser(10, 0))    # lève l'exception
except ValueError as e:
    print("Erreur :", e)     # Erreur : Division par zéro interdite
```

# Ex. 14 (synthèse). Écris somme_fichier(chemin) : le fichier contient un nombre par ligne. La fonction renvoie la somme des nombres valides et ignore les lignes non numériques ou vides. Si le fichier n’existe pas, elle lève FileNotFoundError (laisse-la simplement remonter, sans la capturer).

```python
def somme_fichier(chemin):
    total = 0
    with open(chemin, encoding="utf-8") as fichier:
        for ligne in fichier:
            try:
                total += float(ligne.strip())
            except ValueError:
                pass
    return total

print(somme_fichier("nombres.txt"))   # 17.5
print(somme_fichier("test.txt"))      # 0 (aucune ligne numérique)
```

# Séance 5 : Python « propre » (début de la phase 2)
Les notions

```python
# 1. enumerate : indice + valeur, sans compteur manuel
for i, fruit in enumerate(["pomme", "poire"], start=1):
    print(i, fruit)          # 1 pomme / 2 poire

# 2. zip : parcourir deux listes en parallèle
noms = ["Awa", "Koffi"]
ages = [31, 28]
for nom, age in zip(noms, ages):
    print(nom, age)
dict(zip(noms, ages))        # {"Awa": 31, "Koffi": 28}

# 3. Compréhensions avec condition
[x for x in range(10) if x % 2 == 0]        # [0, 2, 4, 6, 8]
{nom: len(nom) for nom in noms}             # {"Awa": 3, "Koffi": 5}
{x % 3 for x in range(10)}                  # {0, 1, 2}

# 4. any / all et tri
any(x > 5 for x in [1, 2, 9])               # True
all(x > 0 for x in [1, 2, 9])               # True
sorted(["b", "A", "c"], key=str.lower)      # ["A", "b", "c"]
```
